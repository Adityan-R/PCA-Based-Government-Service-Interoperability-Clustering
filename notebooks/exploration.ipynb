## 2. PCA Variance Spectrum & Scree Analysis
We analyze how many principal components are required to capture 90-95% of cumulative variance.

```python
scree_df = pd.read_csv("../data/processed/scree_data.csv")
plt.figure(figsize=(10, 4))
plt.bar(scree_df["component"][:20], scree_df["explained_variance_ratio"][:20] * 100, label="Individual Variance %")
plt.plot(scree_df["component"][:20], scree_df["cumulative_variance_ratio"][:20] * 100, color="red", marker="o", label="Cumulative Variance %")
plt.axhline(92, color="green", linestyle="--", label="92% Variance Threshold")
plt.xlabel("Principal Component Index")
plt.ylabel("Explained Variance (%)")
plt.title("PCA Scree Plot & Cumulative Variance")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
